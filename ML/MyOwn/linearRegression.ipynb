```python
def loss(x,y,m,c):
    n = len(x)
    total_loss = 0
    for i in range(n):
        total_loss += (y[i] - (m*x[i] + c))**2
    return total_loss/n
```

```python
def gradient_descent(x, y, m_curr, c_curr, learning_rate, epochs):
    for _ in range(epochs):
      n = len(x)
      m_gradient = 0
      c_gradient = 0
      for i in range(n):
          m_gradient += (-2/n) * x[i] * (y[i] - (m_curr * x[i] + c_curr))
          c_gradient += (-2/n) * (y[i] - (m_curr * x[i] + c_curr))
      m_new = m_curr - learning_rate * m_gradient
      c_new = c_curr - learning_rate * c_gradient
    return m_new, c_new
    
```

```python
import random
import matplotlib.pyplot as plt
x = [random.uniform(0, 5) for _ in range(100)]
y = [2*xi + 1 + random.uniform(-1, 1) for xi in x]
m = 0
c = 0
learning_rate = 0.053
epochs = 100
m, c = gradient_descent(x, y, m, c, learning_rate, epochs)
print(f"Slope: {m:.5f}, Intercept: {c:.5f}")

print(f"Loss: {loss(x, y, m, c):.5f}")
print(f"line is {m:.5f}x + {c:.5f}")

plt.scatter(x, y, color='blue', label='Data Points')
plt.plot(x, [m*xi + c for xi in x], color='red', label='Regression Line')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Linear Regression')
plt.legend()
plt.show()
```
