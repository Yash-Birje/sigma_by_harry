```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split, cross_val_score, KFold, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
wine = load_wine()
X, y = wine.data, wine.target
```

```python
model = DecisionTreeClassifier(random_state=42)
```

## Part A

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
acc_train_test_split = accuracy_score(y_test, y_pred)
print("Accuracy with Train-Test Split:", acc_train_test_split)
```

## Part B

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
scores_kf = cross_val_score(model, X, y, cv=kf)
print("K-Fold CV Accuracy (mean):", scores_kf.mean())
print('K-fold CV variance:', scores_kf.var())
```

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores_skf = cross_val_score(model, X, y, cv=skf)
print("Stratified K-Fold CV Accuracy (mean):", scores_skf.mean())
print("Stratified K-Fold CV Variance:", scores_skf.var())
```

```python
from sklearn.model_selection import LeaveOneOut
loo = LeaveOneOut()
scores_loo = cross_val_score(model, X, y, cv=loo)
print("LOOCV Accuracy (mean):", scores_loo.mean())
print("LOOCV Variance:", scores_loo.var())
```

## Analysis & Report

1. The Train-Test Split accuracy was **very high (0.96)**, but this may not reflect real-world performance since it depends on one random split.
2. Cross-validation (K-Fold, Stratified K-Fold,LOOCV) gave **lower but more reliable accuracies (~0.87–0.88)**. Which shows our model was more generalized.
3. The variance in CV scores of k-folds was **very less(~0.001-0.002)**, meaning the model performs consistently across different folds. Also it shows that k-fold trained models are more stable as compared to LOOCV trained
4. Conclusion: **Cross-validation is more reliable** for evaluating models compared to a single Train-Test split. Whereas **LOOCV is less stable** as compared to K-fold technique.
